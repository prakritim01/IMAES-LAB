```python
# Experiment 6
# Fuzzy Logic and Constraint Satisfaction

# =========================================================
# PART 1: FUZZY LOGIC - STUDENT PERFORMANCE
# =========================================================

# Triangular membership function
def triangular_membership(x, a, b, c):
    """
    Returns membership value between 0 and 1.
    """

    if x <= a or x >= c:
        return 0.0

    elif x == b:
        return 1.0

    elif x < b:
        return (x - a) / (b - a)

    else:
        return (c - x) / (c - b)


# -----------------------------------------
# Attendance Membership Functions
# -----------------------------------------

def attendance_membership(attendance):

    low = triangular_membership(
        attendance, 0, 0, 60
    )

    medium = triangular_membership(
        attendance, 40, 65, 85
    )

    high = triangular_membership(
        attendance, 70, 100, 100
    )

    return {
        "low": low,
        "medium": medium,
        "high": high
    }


# -----------------------------------------
# Marks Membership Functions
# -----------------------------------------

def marks_membership(marks):

    low = triangular_membership(
        marks, 0, 0, 50
    )

    medium = triangular_membership(
        marks, 40, 60, 80
    )

    high = triangular_membership(
        marks, 70, 100, 100
    )

    return {
        "low": low,
        "medium": medium,
        "high": high
    }


# -----------------------------------------
# Fuzzy Rules
# -----------------------------------------

def fuzzy_rules(attendance, marks):

    attendance_values = attendance_membership(attendance)
    marks_values = marks_membership(marks)

    # Rule 1:
    # IF attendance is low AND marks are low
    # THEN performance is poor

    poor = min(
        attendance_values["low"],
        marks_values["low"]
    )

    # Rule 2:
    # IF attendance is medium AND marks are medium
    # THEN performance is average

    average = min(
        attendance_values["medium"],
        marks_values["medium"]
    )

    # Rule 3:
    # IF attendance is high AND marks are high
    # THEN performance is excellent

    excellent = min(
        attendance_values["high"],
        marks_values["high"]
    )

    # Rule 4:
    # IF attendance is high AND marks are medium
    # THEN performance is good

    good_1 = min(
        attendance_values["high"],
        marks_values["medium"]
    )

    # Rule 5:
    # IF attendance is medium AND marks are high
    # THEN performance is good

    good_2 = min(
        attendance_values["medium"],
        marks_values["high"]
    )

    good = max(good_1, good_2)

    return {
        "Poor": poor,
        "Average": average,
        "Good": good,
        "Excellent": excellent
    }


# -----------------------------------------
# Defuzzification
# -----------------------------------------

def calculate_performance(attendance, marks):

    results = fuzzy_rules(attendance, marks)

    # Output values assigned to each category
    output_values = {
        "Poor": 25,
        "Average": 50,
        "Good": 75,
        "Excellent": 100
    }

    numerator = 0
    denominator = 0

    for category in results:

        membership = results[category]

        numerator += (
            membership *
            output_values[category]
        )

        denominator += membership

    if denominator == 0:
        return 0, results

    performance = numerator / denominator

    return performance, results


# -----------------------------------------
# Test Fuzzy Logic
# -----------------------------------------

def run_fuzzy_system():

    print("\n===================================")
    print("       FUZZY STUDENT PERFORMANCE")
    print("===================================")

    attendance = float(
        input("Enter attendance percentage: ")
    )

    marks = float(
        input("Enter marks percentage: ")
    )

    performance, memberships = calculate_performance(
        attendance,
        marks
    )

    print("\nMembership Values:")

    for category, value in memberships.items():

        print(
            f"{category}: {value:.2f}"
        )

    print(
        f"\nFinal Performance Score: "
        f"{performance:.2f}"
    )

    # Convert numerical output to label

    if performance < 37.5:
        label = "Poor"

    elif performance < 62.5:
        label = "Average"

    elif performance < 87.5:
        label = "Good"

    else:
        label = "Excellent"

    print(
        f"Student Performance: {label}"
    )


# =========================================================
# PART 2: CONSTRAINT SATISFACTION PROBLEM
# TIMETABLE GENERATION
# =========================================================

# Subjects
subjects = [
    "AI",
    "DBMS",
    "OS",
    "CN"
]

# Available time slots
slots = [
    "9-10",
    "10-11",
    "11-12",
    "12-1"
]


# -----------------------------------------
# Check whether assignment is valid
# -----------------------------------------

def is_valid_timetable(timetable, subject, slot):

    # Constraint 1:
    # No two subjects can have the same slot

    if slot in timetable.values():
        return False

    # Constraint 2:
    # AI and DBMS cannot be consecutive
    # in this example.

    if subject == "DBMS" and "AI" in timetable:

        if are_consecutive(
            timetable["AI"],
            slot
        ):
            return False

    if subject == "AI" and "DBMS" in timetable:

        if are_consecutive(
            timetable["DBMS"],
            slot
        ):
            return False

    return True


# -----------------------------------------
# Check consecutive slots
# -----------------------------------------

def are_consecutive(slot1, slot2):

    index1 = slots.index(slot1)
    index2 = slots.index(slot2)

    return abs(index1 - index2) == 1


# -----------------------------------------
# Backtracking CSP Solver
# -----------------------------------------

def solve_timetable(timetable):

    # All subjects assigned
    if len(timetable) == len(subjects):
        return timetable.copy()

    # Select next unassigned subject

    for subject in subjects:

        if subject not in timetable:

            for slot in slots:

                if is_valid_timetable(
                    timetable,
                    subject,
                    slot
                ):

                    timetable[subject] = slot

                    result = solve_timetable(
                        timetable
                    )

                    if result is not None:
                        return result

                    # Backtrack
                    del timetable[subject]

            return None


# -----------------------------------------
# Display timetable
# -----------------------------------------

def run_csp():

    print("\n===================================")
    print("        TIMETABLE CSP")
    print("===================================")

    timetable = solve_timetable({})

    if timetable is None:

        print(
            "No valid timetable exists."
        )

    else:

        print(
            "\nValid timetable generated:"
        )

        for subject, slot in timetable.items():

            print(
                f"{slot} -> {subject}"
            )


# =========================================================
# MAIN
# =========================================================

if __name__ == "__main__":

    run_fuzzy_system()

    run_csp()
```
